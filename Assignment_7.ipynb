## Q. Analytical Derivation

### 1. Show that $x_k^- \sim \mathcal{N}(m_k^-, P_k^-)$

Given the state prediction model:
$$x_k^- = A_{k-1} x_{k-1}^+ + G_{k-1} w_{k-1}$$

Since $x_{k-1}^+ \sim \mathcal{N}(m_{k-1}, P_{k-1})$ and $w_{k-1} \sim \mathcal{N}(0, \Sigma_p)$ are independent Gaussian random variables, their linear combination is also Gaussian.

**Mean ($m_k^-$):**
$$m_k^- = \mathbb{E}[x_k^-] = \mathbb{E}[A_{k-1} x_{k-1}^+ + G_{k-1} w_{k-1}]$$
$$m_k^- = A_{k-1} \mathbb{E}[x_{k-1}^+] + G_{k-1} \mathbb{E}[w_{k-1}]$$
Since $\mathbb{E}[w_{k-1}] = 0$:
$$m_k^- = A_{k-1} m_{k-1}$$

**Covariance ($P_k^-$):**
$$P_k^- = \text{Var}(x_k^-) = \mathbb{E}[(x_k^- - m_k^-)(x_k^- - m_k^-)^T]$$
$$P_k^- = \mathbb{E}[(A_{k-1}(x_{k-1}^+ - m_{k-1}) + G_{k-1}w_{k-1})(A_{k-1}(x_{k-1}^+ - m_{k-1}) + G_{k-1}w_{k-1})^T]$$

Expanding and using the independence of $x_{k-1}^+$ and $w_{k-1}$ (cross-terms become zero):
$$P_k^- = A_{k-1} \mathbb{E}[(x_{k-1}^+ - m_{k-1})(x_{k-1}^+ - m_{k-1})^T] A_{k-1}^T + G_{k-1} \mathbb{E}[w_{k-1} w_{k-1}^T] G_{k-1}^T$$
$$P_k^- = A_{k-1} P_{k-1} A_{k-1}^T + G_{k-1} \Sigma_p G_{k-1}^T$$

---

### 2. Show that $y_k \sim \mathcal{N}(H_k m_k^-, H_k P_k^- H_k^T + \Sigma_m)$

Given the measurement model:
$$y_k = H_k x_k^- + z_k$$
where $z_k \sim \mathcal{N}(0, \Sigma_m)$.

**Mean:**
$$\mathbb{E}[y_k] = \mathbb{E}[H_k x_k^- + z_k] = H_k \mathbb{E}[x_k^-] + \mathbb{E}[z_k] = H_k m_k^-$$

**Covariance:**
$$\text{Var}(y_k) = \mathbb{E}[(y_k - \mathbb{E}[y_k])(y_k - \mathbb{E}[y_k])^T]$$
$$\text{Var}(y_k) = \mathbb{E}[(H_k(x_k^- - m_k^-) + z_k)(H_k(x_k^- - m_k^-) + z_k)^T]$$
Since $x_k^-$ and $z_k$ are independent:
$$\text{Var}(y_k) = H_k \mathbb{E}[(x_k^- - m_k^-)(x_k^- - m_k^-)^T] H_k^T + \mathbb{E}[z_k z_k^T]$$
$$\text{Var}(y_k) = H_k P_k^- H_k^T + \Sigma_m$$

---

### 3. Joint Distribution Proof
Show that:
$$\begin{bmatrix} x_k^- \\ y_k \end{bmatrix} \sim \mathcal{N} \left( \begin{bmatrix} m_k^- \\ H_k m_k^- \end{bmatrix}, \begin{bmatrix} P_k^- & P_k^- H_k^T \\ H_k P_k^- & H_k P_k^- H_k^T + \Sigma_m \end{bmatrix} \right)$$

The joint vector is Gaussian since it is a linear combination of Gaussian vectors. The mean and individual block variances were derived in parts 1 and 2. We only need the cross-covariance:
$$\text{Cov}(x_k^-, y_k) = \mathbb{E}[(x_k^- - m_k^-)(y_k - H_k m_k^-)^T]$$
$$\text{Cov}(x_k^-, y_k) = \mathbb{E}[(x_k^- - m_k^-)(H_k(x_k^- - m_k^-) + z_k)^T]$$
Since $\mathbb{E}[(x_k^- - m_k^-)z_k^T] = 0$:
$$\text{Cov}(x_k^-, y_k) = \mathbb{E}[(x_k^- - m_k^-)(x_k^- - m_k^-)^T] H_k^T = P_k^- H_k^T$$
By symmetry, $\text{Cov}(y_k, x_k^-) = H_k P_k^-$.

---

### 4. Show the updated distribution $x_k^+ \sim \mathcal{N}(m_k, P_k)$

Using the standard **Gaussian Conditioning Theorem**, if:
$$\begin{bmatrix} X \\ Y \end{bmatrix} \sim \mathcal{N} \left( \begin{bmatrix} \mu_X \\ \mu_Y \end{bmatrix}, \begin{bmatrix} \Sigma_{XX} & \Sigma_{XY} \\ \Sigma_{YX} & \Sigma_{YY} \end{bmatrix} \right)$$
Then the conditional distribution is:
$$X \mid Y=y \sim \mathcal{N}(\mu_X + \Sigma_{XY}\Sigma_{YY}^{-1}(y - \mu_Y), \, \Sigma_{XX} - \Sigma_{XY}\Sigma_{YY}^{-1}\Sigma_{YX})$$

Substituting our variables $X = x_k^-$, $Y = y_k$:
* $\mu_X = m_k^-$
* $\mu_Y = H_k m_k^-$
* $\Sigma_{XX} = P_k^-$
* $\Sigma_{XY} = P_k^- H_k^T$
* $\Sigma_{YY} = H_k P_k^- H_k^T + \Sigma_m$

Let the Kalman Gain be defined as:
$$K_k = P_k^- H_k^T (H_k P_k^- H_k^T + \Sigma_m)^{-1}$$

Then:
$$m_k = m_k^- + K_k (y_{obs, k} - H_k m_k^-)$$
$$P_k = P_k^- - K_k H_k P_k^- = (I - K_k H_k)P_k^-$$

---

### 5. Find $\mathbb{E}[x_k^- \mid y_k = y_{obs, k}]$ and $\text{Var}(x_k^- \mid y_k = y_{obs, k})$

By definition, conditioning $x_k^-$ on the realization of $y_k$ yields the posterior parameters $m_k$ and $P_k$ derived in part 4:
$$\mathbb{E}[x_k^- \mid y_k = y_{obs, k}] = m_k = m_k^- + K_k (y_{obs, k} - H_k m_k^-)$$
$$\text{Var}(x_k^- \mid y_k = y_{obs, k}) = P_k = (I - K_k H_k)P_k^-$$

## Q. 1-D Example

### 1. Show $m_k^- = a m_{k-1}$ and $P_k^- = a^2 P_{k-1} + q$
Given $A_{k-1} = a$, $G_{k-1} = 1$, and $\Sigma_p = q$. Substituting these scalar terms into the general equations from Q1:
$$m_k^- = a m_{k-1}$$
$$P_k^- = a P_{k-1} a + 1 \cdot q \cdot 1 = a^2 P_{k-1} + q$$

### 2. Show the update steps
Using scalar variables $H_k = h$ and $\Sigma_m = r$:
The Innovation covariance $S_k$ is:
$$S_k = h^2 P_k^- + r$$
The Kalman gain $K_k$ is:
$$K_k = P_k^- h S_k^{-1} = \frac{P_k^- h}{h^2 P_k^- + r}$$

Thus:
$$m_k = m_k^- + K_k v_k = m_k^- + P_k^- h S_k^{-1} (y_{obs, k} - h m_k^-)$$
$$P_k = (1 - K_k h) P_k^- = (1 - P_k^- h^2 S_k^{-1}) P_k^-$$

### 3. Predictive measurement distribution $p(y_k \mid Y_{k-1})$
Before looking at $y_k$, the distribution follows the prediction step mapping through the measurement function:
$$y_k = h x_k^- + z_k$$
$$\mathbb{E}[y_k \mid Y_{k-1}] = h \mathbb{E}[x_k^- \mid Y_{k-1}] = h m_k^-$$
$$\text{Var}(y_k \mid Y_{k-1}] = h^2 \text{Var}(x_k^-) + \text{Var}(z_k) = h^2 P_k^- + r$$
$$p(y_k \mid Y_{k-1}) = \mathcal{N}(h m_k^-, h^2 P_k^- + r)$$

### 4. Posterior-predictive measurement distribution $p(y_k \mid Y_k)$
After update step $k$, the state distribution updated is $\mathcal{N}(m_k, P_k)$. A future realization of measurement given this knowledge yields:
$$\mathbb{E}[y_k \mid Y_k] = h m_k$$
$$\text{Var}(y_k \mid Y_k) = h^2 P_k + r$$
$$p(y_k \mid Y_k) = \mathcal{N}(h m_k, h^2 P_k + r)$$

## Q. 2D-Position Estimation

### Part A: Matrix Derivations

Given the constant-velocity state configuration:
$$x_k = \begin{bmatrix} p_x(k) \\ p_y(k) \\ v_x(k) \\ v_y(k) \end{bmatrix}$$

From kinematic equations of motion:
$$p_x(k) = p_x(k-1) + v_x(k-1)\Delta t + \frac{1}{2}\Delta t^2 w_x(k-1)$$
$$p_y(k) = p_y(k-1) + v_y(k-1)\Delta t + \frac{1}{2}\Delta t^2 w_y(k-1)$$
$$v_x(k) = v_x(k-1) + \Delta t w_x(k-1)$$
$$v_y(k) = v_y(k-1) + \Delta t w_y(k-1)$$

Rewriting these linear systems in vector-matrix format yields:
$$A = \begin{bmatrix} 1 & 0 & \Delta t & 0 \\ 0 & 1 & 0 & \Delta t \\ 0 & 0 & 1 & 0 \\ 0 & 0 & 0 & 1 \end{bmatrix}, \quad G = \begin{bmatrix} \frac{1}{2}\Delta t^2 & 0 \\ 0 & \frac{1}{2}\Delta t^2 \\ \Delta t & 0 \\ 0 & \Delta t \end{bmatrix}$$

Since only coordinates are measured ($y_k = [p_{meas,x}(k), p_{meas,y}(k)]^T$):
$$H = \begin{bmatrix} 1 & 0 & 0 & 0 \\ 0 & 1 & 0 & 0 \end{bmatrix}$$

```python
Part B: Python implementation

Python
import numpy as np
import matplotlib.pyplot as plt

# 1. Define Simulation Parameters
dt = 1.0  # Time step
N = 50   # Number of steps

# State transition matrices
A = np.array([[1, 0, dt, 0],
              [0, 1, 0, dt],
              [0, 0, 1,  0],
              [0, 0, 0,  1]])

G = np.array([[0.5 * dt**2, 0],
              [0, 0.5 * dt**2],
              [dt, 0],
              [0, dt]])

H = np.array([[1, 0, 0, 0],
              [0, 1, 0, 0]])

# Noise covariances
Sigma_p = np.eye(2) * 0.1  # Process noise covariance
Sigma_m = np.eye(2) * 5.0  # Measurement noise covariance (GPS error)

# Initialize true state and variables
x_true = np.array([[0], [0], [2], [1.5]]) # Initial state (pos_x, pos_y, vel_x, vel_y)
m = np.array([[0], [0], [0], [0]])        # Initial state estimate
P = np.eye(4) * 10.0                       # Initial state uncertainty

# Placeholders for logging
true_positions = []
measured_positions = []
filtered_positions = []

# 2. Execution Loop
for k in range(N):
    # Simulate ground truth dynamics and noisy measurements
    w = np.random.multivariate_normal([0, 0], Sigma_p).reshape(2, 1)
    z = np.random.multivariate_normal([0, 0], Sigma_m).reshape(2, 1)

    x_true = A @ x_true + G @ w
    y_obs = H @ x_true + z

    # --- KALMAN FILTER STEPS ---
    # Prediction Step
    m_minus = A @ m
    P_minus = A @ P @ A.T + G @ Sigma_p @ G.T

    # Update Step
    S = H @ P_minus @ H.T + Sigma_m
    K = P_minus @ H.T @ np.inverse(S) if hasattr(np, 'inverse') else P_minus @ H.T @ np.linalg.inv(S)

    m = m_minus + K @ (y_obs - H @ m_minus)
    P = (np.eye(4) - K @ H) @ P_minus

    # Log values
    true_positions.append(x_true[:2, 0].copy())
    measured_positions.append(y_obs[:, 0].copy())
    filtered_positions.append(m[:2, 0].copy())

# Convert to numpy arrays for plotting
true_positions = np.array(true_positions)
measured_positions = np.array(measured_positions)
filtered_positions = np.array(filtered_positions)

# 3. Plotting Results
plt.figure(figsize=(10, 6))
plt.plot(true_positions[:, 0], true_positions[:, 1], 'g-', label='True Trajectory', linewidth=2)
plt.scatter(measured_positions[:, 0], measured_positions[:, 1], color='red', alpha=0.5, label='Noisy GPS Measurements', s=15)
plt.plot(filtered_positions[:, 0], filtered_positions[:, 1], 'b--', label='Kalman Filter Estimate', linewidth=2)
plt.title('2D Position Estimation via Kalman Filtering')
plt.xlabel('X Position')
plt.ylabel('Y Position')
plt.legend()
plt.grid(True)
plt.show()
```
