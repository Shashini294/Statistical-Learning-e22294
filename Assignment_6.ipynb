1. Feature Selection Strategy and Justification
To predict the predicted_energy_demand within the green building multi-source environmental dataset (2,400 samples), a linear model (Y=Xβ+ϵ) was developed. The selection of input parameters is justified below:

Thermodynamic Causality: Predictors with direct physical ties to building heat transfer equations were prioritized. These include ambient temperature, indoor HVAC temperature setpoints, insulation R-values, and solar radiation levels, as they dictate the continuous energy expenditure required to stabilize the internal climate.

Multicollinearity Mitigation: Environmental variables frequently exhibit high mutual correlation (e.g., relative humidity and dew point temperature). To ensure coefficient stability, a Variance Inflation Factor (VIF) analysis was conducted. Features yielding a VIF>5 were systematically omitted to prevent variance inflation of the optimized parameter weights (β), ensuring that each feature represents an independent source of variance.

Statistical Parsimony: Backed by a robust sample size of 2,400 observations, regularization techniques (such as Lasso or Ridge regression) or stepwise selection methods were utilized to isolate the dominant linear drivers of demand, eliminating uninformative environmental noise and minimizing generalization error.
