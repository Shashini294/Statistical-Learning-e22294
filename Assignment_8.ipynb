### 1. Statistical Mechanics of the Quantum Oscillator

**The Partition Function:**
The partition function is given by the sum:
$Z(T) = \sum_{j=0}^{\infty} e^{-\beta \hbar \omega (j + 1/2)} = e^{-\beta \hbar \omega / 2} \sum_{j=0}^{\infty} (e^{-\beta \hbar \omega})^j$
This is a geometric series $\sum r^j = \frac{1}{1-r}$ with $r = e^{-\beta \hbar \omega}$. Thus:
$Z(T) = \frac{e^{-\frac{\hbar \omega}{2k_BT}}}{1 - e^{-\frac{\hbar \omega}{k_BT}}}$

**Thermodynamic Potentials:**

* **Helmholtz Free Energy:**
$F(T) = -k_B T \ln Z = -k_B T \left[ -\frac{\hbar \omega}{2k_BT} - \ln(1 - e^{-\frac{\hbar \omega}{k_BT}}) \right] = \frac{\hbar \omega}{2} + k_B T \ln(1 - e^{-\frac{\hbar \omega}{k_BT}})$
* **Internal Energy:**
$U(T) = -\frac{\partial \ln Z}{\partial \beta} = \frac{\partial}{\partial \beta} \left[ \frac{\beta \hbar \omega}{2} + \ln(1 - e^{-\beta \hbar \omega}) \right] = \frac{\hbar \omega}{2} + \frac{\hbar \omega e^{-\beta \hbar \omega}}{1 - e^{-\beta \hbar \omega}} = \frac{\hbar \omega}{2} + \frac{\hbar \omega}{e^{\beta \hbar \omega} - 1}$

**Entropy:**
$S = -\frac{\partial F}{\partial T} = k_B \left[ \frac{\hbar \omega}{k_B T(e^{\frac{\hbar \omega}{k_B T}} - 1)} - \ln(1 - e^{-\frac{\hbar \omega}{k_BT}}) \right]$
Verification of $U = F + TS$ confirms the thermodynamic consistency.

**Heat Capacity:**
$C_V = \frac{\partial U}{\partial T} = k_B \left( \frac{\hbar \omega}{2k_B T} \right)^2 \frac{1}{\sinh^2 \left( \frac{\hbar \omega}{2k_B T} \right)}$
*Asymptotic Behavior:* In the high-temperature limit ($k_BT \gg \hbar \omega$), $\sinh(x) \approx x$. Therefore, $C_V \approx k_B \left( \frac{\hbar \omega}{2k_B T} \right)^2 \frac{1}{(\frac{\hbar \omega}{2k_B T})^2} = k_B$. This matches the **classical equipartition theorem** for a 1D harmonic oscillator.

### 2. Maximum Entropy Distribution (Mean and Variance)

To maximize entropy $H(p) = -\int p(x) \ln p(x) dx$ subject to $\int p(x)dx=1$, $\int x p(x)dx=\mu$, and $\int (x-\mu)^2 p(x)dx=\sigma^2$, we use Lagrange multipliers. The solution is the Gaussian distribution:
$p(x) = \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left( -\frac{(x-\mu)^2}{2\sigma^2} \right)$

### 3. Asset Pricing and Portfolio Optimization

Using the Method of Lagrange Multipliers for MaxEnt with constraints on the mean price, mean squared price, and mean log-price:
The PDF takes the form $p(S_T) = \exp(\lambda_0 + \lambda_1 S_T + \lambda_2 S_T^2 + \lambda_3 \ln S_T)$.
This simplifies to:
$p(S_T) = C \cdot S_T^{\lambda_3} \exp(\lambda_1 S_T + \lambda_2 S_T^2)$
where $\lambda_1, \lambda_2, \lambda_3$ are determined numerically by substituting this into the three constraint equations provided for $E[S_T]$, $E[S_T^2]$, and $E[\ln S_T]$.
