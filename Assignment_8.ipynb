## Q1. Statistical Mechanics of the Quantum Oscillator

### 1. The Partition Function

The partition function is given by the sum:
$Z(T) = \sum_{j=0}^{\infty} e^{-\beta E_j} = \sum_{j=0}^{\infty} e^{-\beta \hbar \omega (j + 1/2)}$
where $\beta = \frac{1}{k_B T}$. Factoring out the constant term:
$Z(T) = e^{-\frac{\beta \hbar \omega}{2}} \sum_{j=0}^{\infty} (e^{-\beta \hbar \omega})^j$
This is a convergent geometric series $\sum_{j=0}^{\infty} r^j = \frac{1}{1-r}$ with $r = e^{-\beta \hbar \omega}$. Thus:
$Z(T) = \frac{e^{-\frac{\hbar \omega}{2k_B T}}}{1 - e^{-\frac{\hbar \omega}{k_B T}}}$

### 2. Thermodynamic Potentials

* **Helmholtz Free Energy ($F$):**
$F(T) = -k_B T \ln Z = \frac{\hbar \omega}{2} + k_B T \ln(1 - e^{-\frac{\hbar \omega}{k_B T}})$
* **Internal Energy ($U$):**
$U(T) = -\frac{\partial}{\partial \beta} \ln Z = \frac{\hbar \omega}{2} + \frac{\hbar \omega}{e^{\frac{\hbar \omega}{k_B T}} - 1}$

### 3. Entropy ($S$)

Using $S = -\left(\frac{\partial F}{\partial T}\right)$, we find:
$S(T) = k_B \left[ \frac{\hbar \omega}{k_B T} \frac{1}{e^{\frac{\hbar \omega}{k_B T}} - 1} - \ln(1 - e^{-\frac{\hbar \omega}{k_B T}}) \right]$
Verification: $F + TS = (\frac{\hbar \omega}{2} + k_B T \ln(1 - e^{-\beta \hbar \omega})) + T \cdot S = U$.

### 4. Heat Capacity and Hyperbolic Form

* **Heat Capacity ($C_V$):**
$C_V = \frac{\partial U}{\partial T} = k_B \left( \frac{\hbar \omega}{k_B T} \right)^2 \frac{e^{\frac{\hbar \omega}{k_B T}}}{(e^{\frac{\hbar \omega}{k_B T}} - 1)^2}$
* **Hyperbolic Identity:**
Using $\sinh(x) = \frac{e^x - e^{-x}}{2}$, it simplifies to:
$C_V = k_B \left( \frac{\hbar \omega}{2 k_B T} \right)^2 \frac{1}{\sinh^2\left( \frac{\hbar \omega}{2 k_B T} \right)}$

### 5. Asymptotic Behavior

In the high-temperature limit ($k_B T \gg \hbar \omega$), $\sinh(x) \approx x$. Therefore, $C_V \to k_B$. This matches the **Classical Equipartition Theorem**, which predicts $k_B$ for a 1D harmonic oscillator.

## Q2. The Maximum Entropy Distribution under Mean and Variance Constraints

To maximize entropy $H(p) = -\int p(x) \ln p(x) dx$ subject to constraints $\int p(x)dx=1$, $\int x p(x)dx=\mu$, and $\int (x-\mu)^2 p(x)dx = \sigma^2$, we use the method of Lagrange multipliers. The solution is the Gaussian distribution:
$p(x) = \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left( -\frac{(x-\mu)^2}{2\sigma^2} \right)$

## Q3. Asset Pricing and Portfolio Optimization

To find the Maximum Entropy distribution $p(S_T)$ subject to the constraints on $E[S_T]$, $E[S_T^2]$, and $E[\ln S_T]$, we set up the Lagrangian:
$L = -\int p \ln p \, dS_T - \lambda_0(\int p \, dS_T - 1) - \lambda_1(\int S_T p \, dS_T - 102) - \lambda_2(\int S_T^2 p \, dS_T - 10600) - \lambda_3(\int \ln(S_T) p \, dS_T - 4.60)$
The resulting probability density function takes the form:
$p(S_T) = \frac{1}{Z} \exp\left( -\lambda_1 S_T - \lambda_2 S_T^2 - \lambda_3 \ln(S_T) \right)$
Simplifying the exponential terms:
$p(S_T) = \frac{1}{Z} S_T^{-\lambda_3} \exp\left( -\lambda_1 S_T - \lambda_2 S_T^2 \right)$
where $\lambda_1, \lambda_2, \lambda_3$ are determined by solving the system of equations derived from the provided constraints.
